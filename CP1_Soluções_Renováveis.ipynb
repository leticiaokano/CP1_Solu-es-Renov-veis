# **DATASET 1 - Appliances Energy Prediction (UCI)**

```python
import pandas as pd
```

```python
dataset_1 = pd.read_csv('dados_dataset_1.csv')
```

```python
#Renomeação das Colunas
dataset_1.rename(columns={'Appliances':'Consumo_Eletrodomesticos',
                   'T1':'Temperatura_1',
                   'T2':'Temperatura_2',
                   'T3':'Temperatura_3'
                   }, inplace=True)
dataset_1.head()
```

```python
#Determinando o maior consumo de eletrodomésticos
maior_consumo_1 = dataset_1['Consumo_Eletrodomesticos'].max()

print(f'O maior consumo de eletrodomésticos registrado na amostra foi de {maior_consumo_1}')
```

```python
#Limiar correspondente a 70% do valor máximo e DataFrame contendo somente os registros acima desse limite
limiar_consumo = 0.70 * maior_consumo_1

dataset_acima_limiar = dataset_1[dataset_1['Consumo_Eletrodomesticos'] > limiar_consumo]

print(f'O limiar correspondente a 70% do consumo máximo é de {limiar_consumo}')
dataset_acima_limiar.head()
```

```python
#Registros selecionados e Percentual da amostra
qtd_acima_limiar = len(dataset_acima_limiar)
qtd_total = len(dataset_1)
percentual_acima_limiar = (qtd_acima_limiar / qtd_total) * 100

print(f'Registros acima do limiar: {qtd_acima_limiar}')
print(f'Total de registros na amostra: {qtd_total}')
print(f'Percentual da amostra: {percentual_acima_limiar:.2f}%')
```

```python
#Temperatura média e Segundo dataframe
temperatura_media_1 = dataset_1['Temperatura_1'].mean()

dataset_consumo_e_temperatura = dataset_1[
    (dataset_1['Consumo_Eletrodomesticos'] > limiar_consumo) &
    (dataset_1['Temperatura_1'] > temperatura_media_1)
]

print(f'A temperatura média de T1 na amostra é de {temperatura_media_1:.2f}')
print(f'Registros com consumo acima do limiar E temperatura acima da média: {len(dataset_consumo_e_temperatura)}')
dataset_consumo_e_temperatura.head()
```

```python
qtd_so_consumo = len(dataset_acima_limiar)
qtd_consumo_e_temp = len(dataset_consumo_e_temperatura)
percentual_so_consumo = (qtd_so_consumo / qtd_total) * 100
percentual_consumo_e_temp = (qtd_consumo_e_temp / qtd_total) * 100

print(f'Registros apenas com consumo acima do limiar: {qtd_so_consumo} ({percentual_so_consumo:.2f}% da amostra)')
print(f'Registros com consumo acima do limiar E temperatura acima da média: {qtd_consumo_e_temp} ({percentual_consumo_e_temp:.2f}% da amostra)')
print(f'Redução ao adicionar o segundo critério: {qtd_so_consumo - qtd_consumo_e_temp} registros a menos')
```

# **DATASET 2 - Steel Industry Energy Consumption (UCI)**

```python
import pandas as pd
```

```python
dataset_2 = pd.read_csv('dados_dataset_2.csv')
```

```python
#Renomeação das Colunas
dataset_2.rename(columns={'Usage_kWh':'Consumo_kWh',
                   'Lagging_Current_Reactive.Power_kVarh':'Potencia_reativa_atrasada',
                   'Leading_Current_Reactive_Power_kVarh':'Potencia_reativa_adiantada',
                   'Lagging_Current_Power_Factor':'Fator_potencia_atrasada',
                   'Leading_Current_Power_Factor':'Fator_potencia_adiantada'
                   }, inplace=True)
dataset_2.head()
```

```python
#Inspeção Incial
dataset_2.head()
dataset_2.shape
dataset_2.info()
dataset_2.describe()

```

```python
#Maior consumo e limiar de 75%
maior_consumo_2 = dataset_2['Consumo_kWh'].max()
limiar_consumo_2 = 0.75 * maior_consumo_2

print(f'Maior consumo registrado: {maior_consumo_2} kWh')
print(f'Limiar (75% do máximo): {limiar_consumo_2:.2f} kWh')
```

```python
#DataFrame acima do limiar
dataset_acima_limiar_2 = dataset_2[dataset_2['Consumo_kWh'] > limiar_consumo_2]

qtd_acima_2 = len(dataset_acima_limiar_2)
percentual_acima_2 = qtd_acima_2 / len(dataset_2) * 100

print(f'Registros acima do limiar: {qtd_acima_2} ({percentual_acima_2:.2f}% da amostra)')
```

```python
#Quantos são "Maximum Load"
qtd_maximum_load = (dataset_acima_limiar_2['Load_Type'] == 'Maximum_Load').sum()
print(f'Registros Maximum_Load dentro do filtro: {qtd_maximum_load} de {qtd_acima_2}')
```

```python
#Escolhendo o limite de fator de potência
dataset_2['Fator_potencia_atrasada'].describe()

limite_fp_baixo = dataset_2['Fator_potencia_atrasada'].quantile(0.25)
print(f'Limite escolhido para fator de potência baixo: {limite_fp_baixo:.2f}')
```

```python
#DataFrame com consumo alto E fator de potência baixo

dataset_consumo_fp_baixo = dataset_2[
    (dataset_2['Consumo_kWh'] > limiar_consumo_2) &
    (dataset_2['Fator_potencia_atrasada'] < limite_fp_baixo)
]
print(f'Registros com consumo elevado e fator de potência baixo: {len(dataset_consumo_fp_baixo)}')
```

# **DATASET 3 - Power Consumption of Tetouan City (UCI)**

```python
import pandas as pd
```

```python
dataset_3 = pd.read_csv('dados_dataset_3.csv')
```

```python
# Renomeação das variáveis de consumo
dataset_3.rename(columns={'Temperature':'Consumo_Zona_1',
                           'Humidity':'Consumo_Zona_2',
                           'Wind Speed':'Consumo_Zona_3'
                           }, inplace=True)
dataset_3.head()
```

```python
# Consumo máximo em cada zona
max_zona1 = dataset_3['Consumo_Zona_1'].max()
max_zona2 = dataset_3['Consumo_Zona_2'].max()
max_zona3 = dataset_3['Consumo_Zona_3'].max()

print(f'Máximo Zona 1: {max_zona1}')
print(f'Máximo Zona 2: {max_zona2}')
print(f'Máximo Zona 3: {max_zona3}')
```

```python
# Zona com maior pico de consumo
maximos = {'Consumo_Zona_1': max_zona1, 'Consumo_Zona_2': max_zona2, 'Consumo_Zona_3': max_zona3}
zona_maior_pico = max(maximos, key=maximos.get)

print(f'Zona com maior pico: {zona_maior_pico} ({maximos[zona_maior_pico]})')
```

```python
# Limiar de 70% e DataFrame filtrado
maior_consumo_3 = maximos[zona_maior_pico]
limiar_consumo_3 = 0.70 * maior_consumo_3

dataset_acima_limiar_3 = dataset_3[dataset_3[zona_maior_pico] > limiar_consumo_3]

print(f'Limiar (70% do máximo da {zona_maior_pico}): {limiar_consumo_3:.2f}')
dataset_acima_limiar_3.head()
```

```python
# Contagem e percentual
qtd_acima_3 = len(dataset_acima_limiar_3)
percentual_acima_3 = qtd_acima_3 / len(dataset_3) * 100

print(f'Registros acima do limiar: {qtd_acima_3}')
print(f'Total de registros na amostra: {len(dataset_3)}')
print(f'Percentual da amostra: {percentual_acima_3:.2f}%')
```

```python
# Temperatura média e segundo DataFrame
temperatura_media_3 = dataset_3['Consumo_Zona_1'].mean()

dataset_consumo_e_temp_3 = dataset_3[
    (dataset_3[zona_maior_pico] > limiar_consumo_3) &
    (dataset_3['Consumo_Zona_1'] > temperatura_media_3)
]

print(f'Temperatura média (Consumo_Zona_1): {temperatura_media_3:.2f}')
print(f'Registros com os dois critérios: {len(dataset_consumo_e_temp_3)}')
dataset_consumo_e_temp_3.head()
```

```python
# Comparação entre os dois conjuntos
percentual_dois_criterios = len(dataset_consumo_e_temp_3) / len(dataset_3) * 100

print(f'Só consumo acima do limiar: {qtd_acima_3} ({percentual_acima_3:.2f}% da amostra)')
print(f'Consumo + temperatura: {len(dataset_consumo_e_temp_3)} ({percentual_dois_criterios:.2f}% da amostra)')
print(f'Redução ao adicionar o segundo critério: {qtd_acima_3 - len(dataset_consumo_e_temp_3)} registros a menos')
```

# **DATASET 5 - Wind & Solar Energy Production (Kanggle)**

```python
import pandas as pd
```

```python
dataset_5 = pd.read_csv('dados_dataset_5.csv')
```

```python
# 1. Renomeação da variável principal e separação por fonte
dataset_5.rename(columns={'Production':'Geracao'}, inplace=True)

geracao_solar = dataset_5[dataset_5['Source'] == 'Solar']['Geracao']
geracao_eolica = dataset_5[dataset_5['Source'] == 'Wind']['Geracao']
```

```python
# 2. Valor máximo registrado para cada fonte
maior_geracao_solar = geracao_solar.max()
maior_geracao_eolica = geracao_eolica.max()

print(f'Maior geração solar registrada: {maior_geracao_solar}')
print(f'Maior geração eólica registrada: {maior_geracao_eolica}')
```

```python
# 3. Limiar de 70% do máximo de cada fonte
limiar_solar = 0.70 * maior_geracao_solar
limiar_eolica = 0.70 * maior_geracao_eolica

print(f'Limiar (70% do máximo solar): {limiar_solar:.2f}')
print(f'Limiar (70% do máximo eólico): {limiar_eolica:.2f}')
```

```python
# 4. DataFrames de alta geração para cada fonte
dataset_alta_geracao_solar = dataset_5[
    (dataset_5['Source'] == 'Solar') & (dataset_5['Geracao'] > limiar_solar)
]

dataset_alta_geracao_eolica = dataset_5[
    (dataset_5['Source'] == 'Wind') & (dataset_5['Geracao'] > limiar_eolica)
]
```

```python
# 5. Contagem e percentual de cada DataFrame
qtd_alta_solar = len(dataset_alta_geracao_solar)
qtd_alta_eolica = len(dataset_alta_geracao_eolica)

percentual_alta_solar = qtd_alta_solar / len(geracao_solar) * 100
percentual_alta_eolica = qtd_alta_eolica / len(geracao_eolica) * 100

print(f'Registros de alta geração solar: {qtd_alta_solar} ({percentual_alta_solar:.2f}% dos registros solares)')
print(f'Registros de alta geração eólica: {qtd_alta_eolica} ({percentual_alta_eolica:.2f}% dos registros eólicos)')
```

```python
# 6. Comparação entre as fontes
if percentual_alta_solar > percentual_alta_eolica:
    fonte_maior_frequencia = 'Solar'
elif percentual_alta_eolica > percentual_alta_solar:
    fonte_maior_frequencia = 'Eólica'
else:
    fonte_maior_frequencia = 'Empate'

print(f'Fonte com maior frequência acima de 70% do próprio máximo: {fonte_maior_frequencia}')
```

# **DATASET 6 - Individual Household Electric Power Consumption (UCI)**

```python
import pandas as pd
```

```python
dataset_6 = pd.read_csv('dados_dataset_6.csv')
```

```python
# 1. Renomeação das Colunas
dataset_6.rename(columns={'Global_active_power':'Potencia_Ativa',
                           'Global_intensity':'Corrente'
                           }, inplace=True)
```

```python
# 2. Maior potência ativa
maior_potencia_ativa = dataset_6['Potencia_Ativa'].max()
print(f'A maior potência ativa registrada na amostra foi de {maior_potencia_ativa}')
```

```python
# 3. Limiar de 75% e DataFrame filtrado
limiar_potencia_ativa = 0.75 * maior_potencia_ativa
dataset_acima_limiar_6 = dataset_6[dataset_6['Potencia_Ativa'] > limiar_potencia_ativa]

print(f'Limiar (75% da potência máxima): {limiar_potencia_ativa:.2f}')
```

```python
# 4. Quantidade e percentual
qtd_acima_6 = len(dataset_acima_limiar_6)
percentual_acima_6 = qtd_acima_6 / len(dataset_6) * 100

print(f'Registros acima do limiar: {qtd_acima_6}')
print(f'Percentual da amostra: {percentual_acima_6:.4f}%')
```

```python
# 5. Corrente média
corrente_media = dataset_6['Corrente'].mean()
print(f'Corrente média da amostra: {corrente_media:.2f} A')
```

```python
# 6. Segundo DataFrame com os dois critérios
dataset_potencia_e_corrente = dataset_6[
    (dataset_6['Potencia_Ativa'] > limiar_potencia_ativa) &
    (dataset_6['Corrente'] > corrente_media)
]

print(f'Registros com potência elevada e corrente acima da média: {len(dataset_potencia_e_corrente)}')
```

```python
# 7. Comparação
percentual_dois_criterios = len(dataset_potencia_e_corrente) / len(dataset_6) * 100

print(f'Só potência ativa acima do limiar: {qtd_acima_6} ({percentual_acima_6:.4f}%)')
print(f'Potência + corrente: {len(dataset_potencia_e_corrente)} ({percentual_dois_criterios:.4f}%)')
```
